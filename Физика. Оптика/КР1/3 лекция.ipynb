Интенсивность - модуль среднего значения по времени вектора Поинтинга
$$
I = \left|<\vec S>\right| = \left|<\vec E\times\vec H>\right| = \frac{1}2\sqrt{\frac{\varepsilon\varepsilon_0}{\mu\mu_0}}E^2_m,\,\text{где}\, E_m = A
$$
Упрощается за счёт среды и $\mu\approx 1$ :
$$
I = \frac{1}2\sqrt{\varepsilon}E^2_m = nA^2
$$
Для некогерентных волн сложение выглядит так :
$$
I = I_1 + I_2
$$
Для когерентных волн сложение выглядит так :
$$
I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos(\varphi_2 - \varphi_1)
$$
При максимальной интенсивности косинус будет обращаться в положительную единицу, при минимальной в отрицательную, а значит :
$$
\begin{aligned}
I_{max}=I_1 + I_2 + 2\sqrt{I_1I_2}
\\
I_{min}=I_1 + I_2 - 2\sqrt{I_1I_2}
\end{aligned}
$$
В свою очередь если амплитуды источников равны, то картина слудующая :
$$
\begin{aligned}
I_{max}=4I
\\
I_{min}=0
\end{aligned}
$$
В провялении таких минимумов и максимумов и заключается явление интерференции

Волны можно представить как :
$$
E_1 = a_1\cos(\omega\frac{S_1}v_1)
\\
E_2 = a_2\cos(\omega\frac{S_2}v_2)
$$
Механизм интерференции же :
$$
\begin{cases}
E_1 = a_1\cos(\alpha_1)
\\
E_2 = a_2\cos(\alpha_2)
\\
I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos{(\alpha_2 - \alpha_1)}
\\
I_{max} : \cos{(\alpha_2 - \alpha_1)} = 1
\\
I_{min} : \cos{(\alpha_2 - \alpha_1)} = -1
\end{cases}
$$
Из этого следует, что :
$$
\begin{cases}
\alpha_1 = \omega\frac{S_1}{v_1}
\\
\alpha_2 = \omega\frac{S_2}{v_2}
\\
I_{max} : \omega(\frac{S_2}{v_2} - \frac{S_1}{v_1}) = 0
\\
I_{min} : \omega(\frac{S_2}{v_2} - \frac{S_1}{v_1}) = \pi
\end{cases}
$$
Из чего также выводится :
$$
\alpha_2 - \alpha_1 = \frac{2\pi}{\lambda_0}(S_2n_2-S_1n_1)
$$
Также верно и следующее из Т Пифагора :
$$
S^2_2 - S^2_1 = (S_1 + S_2)(S_2 - S_1)
\\
S^2_1 = l^2 + (x + \frac{D}2)^2\qquad S^2_2 = l^2 + (x - \frac{D}2)^2
\\
S_2 - S_1 = \frac{2Dx}{S_2+S_1}
$$
Переход к условию максимума интенсивности :
$$
\Delta S \approx\frac{D}lx
\\
\frac{2\pi}\lambda\Delta S = 2\pi m
\\
x_{max}=\frac{l}Dm\lambda,\,\text{где}\,m=0,\pm 1,\pm 2,\,...
$$
