Интенсивность - модуль среднего значения по времени вектора Поинтинга
$$
I = |<\vec S>| = |<\vec E\times\vec H>| = \frac{1}2\sqrt{\frac{\varepsilon\varepsilon_0}{\mu\mu_0}}E^2_m,\,\text{где}\, E_m = A
$$
Упрощается за счёт среды и $\mu\approx 1$ :
$$
I = \frac{1}2\sqrt{\varepsilon}E^2_m = nA^2
$$
Для некогерентных волн сложение выглядит так :
$$
I = I_1 + I_2
$$
Для когерентных волн сложение выглядит так :
$$
I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos(\varphi_2 - \varphi_1)
$$
При максимальной интенсивности косинус будет обращаться в положительную единицу, при минимальной в отрицательную, а значит :
$$
\begin{aligned}
I_{max}=I_1 + I_2 + 2\sqrt{I_1I_2}
\\
I_{min}=I_1 + I_2 - 2\sqrt{I_1I_2}
\end{aligned}
$$
В свою очередь если амплитуды источников равны, то картина слудующая :
$$
\begin{aligned}
I_{max}=4I
\\
I_{min}=0
\end{aligned}
$$
В провялении таких минимумов и максимумов и заключается явление интерференции
