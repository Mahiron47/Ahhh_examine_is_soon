Волна - это возмущение среды, переносящее энергию, частицы среды не движутся вместе с волной, а колеблются около своих положений равновесия

Продольная волна - волна, колебания частиц которой происходят вдоль плоскости распространения волны

Поперечная волна - волна, колебания частиц которой происходят вдоль ортогональной плоскости по отношению к плоскости распространения волны

Уравнение волны описывается в координатах $(x,\,y,\,z,\,t)$ :
$$
\xi(x,\,t) = A\cos(\omega t - kx + \varphi_0)\,,\text{ где }\, \omega=2\pi\nu
$$
Скорость волны $v(\omega)$ :
$$
v(\omega)=\frac{\omega}k\,,\quad k\text{ - волновове число },\quad\lambda=\frac{2\pi}k\text{ - длинна волны}
$$

Волна в поглащающей среде вдоль оси $x$ :
$$
\xi(x, t)=a_0e^{-\gamma x}\cos(\omega t - kx + \varphi_0)
$$
Уравнение плоской волны распространяющейся в трёхмерном пространстве :
$$
\xi(\vec r, t)=A\cos(\omega t - \vec k\cdot\vec r + \varphi_0)\quad k=\frac{2\pi}\lambda
$$
Уравнение сферической волны, распространяющейся в трёхмерном пространстве :
$$
\xi(\vec r, t)=\frac{a_0}r\cos(\omega t - kr + \varphi_0)
$$
