Электромагнитные волны
</span>
</center>

Поток фотонов описывается из уравнений Максвелла электромагнитными волнами : 
$$
\begin{aligned}
\nabla^2\vec E=\varepsilon\varepsilon_0\mu\mu_0\frac{\delta^2\vec E}{\delta t^2}
\\
\nabla^2\vec H=\varepsilon\varepsilon_0\mu\mu_0\frac{\delta^2\vec H}{\delta t^2}
\end{aligned}
$$
Скорость распространения электромагнитной волны $v$ :
$$
\begin{aligned}
v=\frac{c}{\sqrt{\varepsilon\mu}}
\\
c=\frac{1}{\sqrt{\varepsilon_0\mu_0}}
\end{aligned}
$$
Векторы $\vec E$, $\vec H$ и $\vec k$ образуют правую тройку

Решениями уравнений же будет :
$$
\begin{aligned}
\vec E = E_0\cos(\omega t - \vec k\cdot\vec r_1 + \varphi_1)
\\
\vec H = H_0\cos(\omega t - \vec k\cdot\vec r_2 + \varphi_2)
\\
\varphi_1=\varphi_2=\varphi
\end{aligned}
$$

Мгновенные значения $E$ и $H$ связаны следующим уравнением :
$$
\sqrt{\varepsilon\varepsilon_0}E=\sqrt{\mu\mu_0}H
$$
Вектор пойнтинга указывает направление переноса энергии и сонавправлен с вектором $\vec k$, а равен ежесекундному переносу энергии через единичную площадку :
$$
\vec S = \vec E\times\vec H
$$
Объёмная плотность энергии :
$$
w=\frac{\varepsilon\varepsilon_0E^2}2+\frac{\varepsilon\varepsilon_0H^2}2
$$
Вектор пойнтинга из объёмной плотности энергии :
$$
\vec S = w\vec v
$$
