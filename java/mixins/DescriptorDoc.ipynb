Дескриптор поля в JVM включает :
- Дескриптор класса, в котором объявлено поле.
- Имя поля : указывается сразу после дескриптора класса + ':' на конце.
- Дескриптор типа поля : указывается сразу после ':'.

Дескриптор метода в JVM включает :
- Дескриптор класса, в котором объявлен метод.
- Имя метода : указывается сразу после дескриптора класса, имя конструктора - `<init>`, имя статического конструктора - `<clinit>`, второе имя любого метода - `<method>`.
- Дескриптор аргументов : после имени в круглых скобках () перечисляются типы параметров.
- Дескриптор типа возвращаемого значения : указывается после скобок.

Дескрипторы базовых типов JVM и их обозначения :
| Символ | Тип Java       |
| :----- | :------------- |
| `B`    | `byte`         |
| `C`    | `char`         |
| `D`    | `double`       |
| `F`    | `float`        |
| `I`    | `int`          |
| `J`    | `long`         |
| `L;`   | Ссылочный тип  |
| `S`    | `short`        |
| `Z`    | `boolean`      |
| `[`    | Массив         |

Именования дескриптора поля `private static Object field` класса `net.example.constants` JVM :
```java
Lnet/example/constants;field:Ljava/lang/Object;
```
Именование дескриптора нулевого (базового) конструктора JVM :
```java
<init>()V
```
Именование дескриптора нулевого (базового) статического конструктора JVM :
```java
<clinit>()V
```
Именование дескриптора метода `Object create(int i, double[] d, Thread t)` JVM :
```java
create(I[DLjava/lang/Thread;)Ljava/lang/Object;
```
Если метод `Object create(int i, double[] d, Thread t)` находится в классе `net.example.methods` то дескриптор будет включать дескриптор класса :
```java
Lnet/example/methods;create(I[DLjava/lang/Thread;)Ljava/lang/Object;
```
