<center style="font-size:2em;">Общая справка</center>

Мод **Computer Craft** вводит компьютеры работающие на **DOS** подобной системе, и соответственно взаимодействие с файлами верное для консоли **Windows**, также верно и для **CraftOS**.

Изначально на каждом устройстве содержится одна директория только для чтения - **rom**. Она содержит весь код доступный для использования в своём коде.

Для взаимодействия с внешним миром, устройство, внутри пользовательского кода, должно вызывать инструкции, поставляемые **API**.

Пользовательский код, записанный в файл, может быть исполнен устройством, путём написания его названия заместо стандартной инструкции внутри одной с файлом директории.

<center style="font-size:2em;">Инструкция ls или list</center>

Команда **ls** используется для отображения списка файлов и директорий в текущей директории.

<center>
ls
</center>


<center style="font-size:2em;">Инструкция mkdir</center>

Команда **mkdir** (make directory) используется для создания новой директории.
<center>
mkdir <имя_директории>
</center>

<center style="font-size:2em;">Инструкция cd</center>

Команда **cd** (change directory) используется для смены текущей директории.

<center>
cd <путь_к_директории>
</center>

также

<center>
cd .
</center>

обозначает перемещение в текущую директорию, увеличение количества точек указывает на переход на уровень выше.

<center>
cd \
</center>

перемещает на корневую директорию.

<center style="font-size:2em;">Инструкция mv</center>

Команда **mv** (move) используется для перемещения или переименования файлов и директорий.

<center>
mv <имя_файла_или_директории> <новое_имя>
</center>

также 

<center>
mv <имя_файла_или_директории> <путь_к_новому_месту>
</center>

перемещает файл или директорию в новое место.

<center style="font-size:2em;">Инструкция cp или copy</center>

Команда **cp** (copy) используется для копирования файлов и директорий.

<center>
cp <имя_файла_или_директории> <имя_копии>
</center>

также

<center>
cp <имя_файла_или_директории> <путь_к_новому_месту>
</center>

копирует файл или директорию в новое место с тем же именем.

<center style="font-size:2em;">Инструкция rm или delete</center>

Команда **rm** (remove) используется для удаления файлов и директорий.

<center>
rm <имя_файла_или_директории>
</center>

<center style="font-size:2em;">Инструкция clear</center>

Команда **clear** используется для очистки экрана терминала.

<center>
clear
</center>

<center style="font-size:2em;">Инструкция type</center>

Команда **type** используется для получения информации о типе объекта.

<center>
type <имя_объекта>
</center>


<center style="font-size:2em;">Инструкция apis</center>

Команда **apis** используется для отображения списка доступных API.

<center>
apis
</center>


<center style="font-size:2em;">Инструкция edit</center>

Команда **edit** используется для редактирования текстовых файлов.

<center>
edit <имя_файла>
</center>
